### 4.1 Purpose of the Literature Review

### 4.2 Exploratory Data Analysis and the Data Science Workflow

IBM's documentation on CRISP-DM describes a data-mining workflow that includes business understanding, data understanding, data preparation, modeling, evaluation, and deployment. It also explains that these phases are iterative rather than strictly linear.

This framework supports the approach taken in my project: first understanding the dataset, examining its structure and quality, investigating variable distributions and relationships, and then identifying considerations for subsequent model development.

**Relevance to my project:** I used EDA to understand the available columns, identify missing values and duplicate records, inspect numerical and categorical distributions, and explore relationships between variables. This is consistent with the data-understanding and data-preparation stages of the CRISP-DM framework.

### 4.3 Data Preparation and Data Leakage

The scikit-learn documentation, *Common pitfalls and recommended practices*, explains that data leakage occurs when information unavailable at prediction time is used during model development. It recommends splitting data into training and testing sets before fitting preprocessing steps and using pipelines to keep transformations consistent.

**Relevance to my project:** Missing values and categorical variables require appropriate preprocessing before model training. Imputation, encoding, and other transformations that learn from data should be fitted on the training set and then applied to the test set. EDA can still identify missing values and describe their patterns before splitting; the important distinction is between inspecting missingness and fitting a data-dependent imputation strategy.

The documentation also highlights the need to consider whether a feature would genuinely be available at the intended prediction time. For example, delivery information, customer reviews, ratings, and order-status fields may contain information generated after an order is placed or returned. Their suitability depends on the exact prediction scenario.

### 4.4 Evaluation of Imbalanced Classification Problems

Google for Developers explains that accuracy alone can be misleading when one class is much more common than another. Precision measures how many predicted positive cases are actually positive, while recall measures how many actual positive cases the model identifies. F1 combines precision and recall.

The imbalanced-learn documentation provides tools and guidance for working with datasets in which class frequencies differ substantially.

**Relevance to my project:** The observed return rate in the cleaned dataset is approximately 7.01%, meaning that non-returned orders are much more common than returned orders. Therefore, a model should not be judged by accuracy alone. Precision, recall, F1-score, the confusion matrix, and precision-recall analysis are useful for understanding how effectively the model identifies returned orders.

These sources establish general evaluation principles; they do not demonstrate that a particular model will perform well on this synthetic dataset.

### 4.5 Scope and Limitations of the Review

This is a focused review of technical documentation, not a systematic literature review or a comprehensive review of peer-reviewed research on e-commerce returns. The sources were selected because they directly address the project's analytical workflow and machine-learning methodology.

No claim is made that these references validate the dataset's synthetic patterns, explain the causes of returns in real marketplaces, or establish causal relationships between discounts, ratings, delivery times, and returns.


## 5. What I Learned from the Literature Review

The technical documentation helped me establish the following principles for my project:

1. **Follow a structured workflow:** Understand the business problem and dataset before moving to preprocessing, model development, and evaluation.
2. **Separate data inspection from preprocessing:** Missing values can be identified and studied during EDA, but imputation parameters should be learned from training data only.
3. **Prevent data leakage:** Features and preprocessing steps must not expose information from the test set or from events that occur after the intended prediction time.
4. **Choose features according to the prediction scenario:** A variable can be available in a completed-order dataset but still be unsuitable for predicting a return before that information exists.
5. **Use more than accuracy:** Because returned orders are the minority class, precision, recall, F1-score, and the confusion matrix are important complementary evaluation measures.
6. **Interpret results within their limitations:** Patterns found in a synthetic dataset describe that dataset. They cannot automatically be generalized to actual customers or marketplace operations.
7. **Distinguish association from causation:** A relationship observed in a chart does not, by itself, prove that one variable causes another.

These are methodological lessons drawn from the cited documentation and applied to the decisions made during this project. They are not claims that the cited sources independently tested my dataset or confirmed my results.

### Important feature-selection note

Although all 30 columns are useful for understanding the dataset, they should not all automatically be used as model features.

`returned_flag` is the target, and `returned` is another representation of the same outcome. `order_status` may directly reveal whether the order was returned. These columns should be excluded from the predictor features.

Identifiers such as `order_id`, `customer_id`, and `product_id` generally should not be treated as ordinary numerical predictors. Date columns may require feature engineering, and their suitability depends on the prediction scenario.

Review information, customer ratings, and delivery information also require special consideration: they may only become available after the purchase or delivery. If the intended model predicts returns before those events, these features would not be available at prediction time.

### 8.1 Role of LLMs in Dataset Creation

### 8.2 How LLMs Were Used in This Project

LLMs were used as an assistance tool during the project for tasks such as:

- **Synthetic data generation:** Assisting in creating a structured dataset that resembles an e-commerce marketplace scenario.
- **Analytical guidance:** Helping identify appropriate EDA techniques for numerical and categorical variables.

LLM-generated explanations and suggestions were treated as assistance rather than authoritative evidence. Statistical findings were derived from analysis of the dataset, and methodological recommendations were checked against technical documentation where applicable.

### 8.3 Limitations and Risks of LLM-Generated Data

Synthetic data introduces important limitations:

1. **No direct real-world validation:** The dataset does not establish how actual Amazon customers behave or why real orders are returned.
2. **Potential artificial patterns:** Distributions, correlations, and category differences may arise from the generation process rather than real marketplace conditions.
3. **Potential internal inconsistencies:** Generated records may contain implausible values, duplicated patterns, or relationships that do not reflect real business processes.
4. **Model evaluation limitations:** A model trained and tested on this dataset measures its performance on the generated data distribution. Its performance cannot be assumed to transfer to real-world e-commerce data.
5. **Limited causal interpretation:** Observed relationships between variables cannot be used to establish causal explanations for returns.

Consequently, any findings from this project are described as patterns observed in the synthetic dataset, rather than facts about Amazon's actual marketplace.

### 8.4 Responsible Use of LLMs

This project demonstrates the use of LLMs as tools for learning, experimentation, and analytical assistance. However, the use of an LLM does not replace data validation, statistical reasoning, source verification, or critical evaluation.

The analysis, visualizations, and reported statistics should be reproducible from the dataset and code. Any AI-assisted explanations should be reviewed before being accepted as conclusions.
